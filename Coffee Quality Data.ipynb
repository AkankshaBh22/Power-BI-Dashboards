```python
nums = [2,3,4,3,2]
def removeElement(nums, val):
    # Initialize two pointers, one for iterating the array and one for tracking the index of non-val elements
    i = 0
    k = 0
    
    # Iterate through the array
    while i < len(nums):
        # If the current element is not equal to val
        if nums[i] != val:
            # Swap the current element with the element at index k
            nums[i], nums[k] = nums[k], nums[i]
            # Increment the index of non-val elements
            k += 1
        
        # Move to the next element in the array
        i += 1
    
    # Return the number of non-val elements
    return k
```
